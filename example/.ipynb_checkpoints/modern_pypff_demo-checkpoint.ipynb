# Modernized pypff Demo

This notebook demonstrates the new high-performance, zero-copy features of the modernized `pypff` package. 

### Key Concepts:
- **`PanosetiRun`**: Scans a directory for data products and configuration files.
- **`PFFSequence`**: Manages a sequence of PFF files as a single contiguous stream.
- **Pydantic Models**: Provides strictly-typed headers and configurations.

```python
import os
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from pypff import PanosetiRun, PFFSequence, hkpff

# Set path to example data
example_data_dir = Path("./example-data")
```

## 1. Run Discovery

Use `PanosetiRun` to scan an entire observing directory. It automatically groups PFF files into data products.

```python
run = PanosetiRun(example_data_dir)
print(f"Loaded run: {run.run_dir.name}")
print("\nAvailable Data Products:")
for product in run.list_products():
    seq = run.get_product(product)
    print(f" - {product}: {len(seq)} frames")
```

### Accessing Run Configuration
Configuration files in the directory are loaded as parsed dictionaries or models.

```python
print(f"Observatory Name: {run.configs['obs_config']['name']}")
print(f"Run Type: {run.configs['data_config']['run_type']}")
```

## 2. High-Performance Data Access

Let's look at a Pulse Height (PH) product. We use `PFFSequence` for zero-copy access via memory-mapping.

```python
ph_seq = run.get_product("dp_ph256.bpp_2.module_254")

# Get a single frame (header + image)
header, image = ph_seq.get_frame(0)

print("Metadata for Frame 0:")
print(header)
print(f"\nPrecise Timestamp: {header.timestamp_ns} ns")
```

### Zero-Copy Virtual Arrays

The `get_image_array` method is highly optimized. If the file structure allows, it returns a zero-copy strided view of the memory-mapped file, making it extremely fast for reading large slices of data.

```python
# Fetch 100 frames as a single NumPy array (count, H, W)
images = ph_seq.get_image_array(start=0, count=100)
print(f"Fetched slice shape: {images.shape}")
print(f"Array base is the mmap: {images.base is not None}")
```

## 3. Visualization

Let's plot some data from different modes.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# PH256 (16x16)
_, img_ph = run.get_product("dp_ph256.bpp_2.module_254").get_frame(0)
axes[0].imshow(img_ph, cmap='plasma')
axes[0].set_title("PH256 (16x16)")

# IMG16 (32x32)
_, img_movie = run.get_product("dp_img16.bpp_2.module_1").get_frame(0)
axes[1].imshow(img_movie, cmap='viridis')
axes[1].set_title("IMG16 (32x32)")

plt.show()
```

## 4. Housekeeping Data

Housekeeping data is often large and JSON-based. The modernized `hkpff` streams the file to extract data efficiently.

```python
hk = hkpff(example_data_dir / "hk.pff").readhk()

if "QUABO_1019" in hk:
    temps = hk["QUABO_1019"]["DET_TEMP"]
    plt.figure(figsize=(10, 4))
    plt.plot(temps, marker='.')
    plt.title("Detector Temperature Over Time (QUABO 1019)")
    plt.ylabel("Temp (C)")
    plt.xlabel("Sample Index")
    plt.grid(True)
    plt.show()
```
